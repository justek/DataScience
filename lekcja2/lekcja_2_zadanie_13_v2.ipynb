```python
import os
import subprocess
import json
```

```python
def setup_project(project_name):

    print("===== PRZYGOTOWANIE PROJEKTU =====")

    # Tworzenie głównego folderu
    os.makedirs(project_name, exist_ok=True)

    # Struktura projektu
    folders = [
        "data/raw",
        "data/processed",
        "notebooks",
        "src",
        "models",
        "reports"
    ]

    for folder in folders:
        os.makedirs(os.path.join(project_name, folder), exist_ok=True)

    print("Foldery zostały utworzone.")

    # Repozytorium Git
    git_folder = os.path.join(project_name, ".git")

    if not os.path.exists(git_folder):
        subprocess.run(["git", "init"], cwd=project_name)
        print("Repozytorium Git zostało zainicjalizowane.")
    else:
        print("Repozytorium Git już istnieje.")

    # Środowisko wirtualne
    venv_folder = os.path.join(project_name, "venv")

    if not os.path.exists(venv_folder):
        subprocess.run(["python3", "-m", "venv", "venv"], cwd=project_name)
        print("Środowisko wirtualne zostało utworzone.")
    else:
        print("Środowisko wirtualne już istnieje.")

    # Instalacja bibliotek
    pip_path = os.path.join(project_name, "venv", "bin", "pip")

    if os.path.exists(pip_path):
        subprocess.run([
            pip_path,
            "install",
            "jupyter",
            "pandas",
            "numpy",
            "matplotlib"
        ])
        print("Podstawowe biblioteki zostały zainstalowane.")
    else:
        print("Nie można zainstalować bibliotek - nie znaleziono środowiska wirtualnego.")

    # README.md
    readme_path = os.path.join(project_name, "README.md")

    if not os.path.exists(readme_path):
        with open(readme_path, "w") as f:
            f.write(f"# {project_name}\n\n")
            f.write("Projekt Data Science.\n")
        print("Plik README.md został utworzony.")
    else:
        print("Plik README.md już istnieje.")

    # .gitignore
    gitignore_path = os.path.join(project_name, ".gitignore")

    if not os.path.exists(gitignore_path):
        with open(gitignore_path, "w") as f:
            f.write(".DS_Store\n")
            f.write(".ipynb_checkpoints/\n")
            f.write("venv/\n")
            f.write("__pycache__/\n")
            f.write("data/raw/\n")
        print("Plik .gitignore został utworzony.")
    else:
        print("Plik .gitignore już istnieje.")

    # Szablon notebooka
    notebook_path = os.path.join(
        project_name,
        "notebooks",
        "template.ipynb"
    )

    if not os.path.exists(notebook_path):

        notebook = {
            "cells": [],
            "metadata": {},
            "nbformat": 4,
            "nbformat_minor": 5
        }

        with open(notebook_path, "w") as f:
            json.dump(notebook, f, indent=2)

        print("Szablon notebooka został utworzony.")
    else:
        print("Szablon notebooka już istnieje.")

    print("\n===== PRZYGOTOWANIE PROJEKTU ZAKOŃCZONE =====")
```

```python
setup_project("SalesAnalysis")
```
