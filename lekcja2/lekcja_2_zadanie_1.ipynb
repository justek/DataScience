```python
# Sprawdźmy, jakie zasoby są dostępne w Google Colab
import sys
import psutil
import platform
# Informacje o systemie
print("=== INFORMACJE O SYSTEMIE ===")
print(f"System operacyjny: {platform.system()} {platform.release()}")
print(f"Python version: {sys.version}")
print()
# Informacje o pamięci RAM
print("=== PAMIĘĆ RAM ===")
ram_gb = psutil.virtual_memory().total / (1024**3)
ram_available_gb = psutil.virtual_memory().available / (1024**3)
print(f"Całkowita RAM: {ram_gb:.2f} GB")
print(f"Dostępna RAM: {ram_available_gb:.2f} GB")
print()
# Informacje o CPU
print("=== PROCESOR ===")
print(f"Liczba rdzeni CPU: {psutil.cpu_count()}")
print(f"Obciążenie CPU: {psutil.cpu_percent(interval=1)}%")
print()
# Sprawdzenie dostępności GPU (w Colab)
try:
    import torch
    if torch.cuda.is_available():
        print("=== GPU ===")
        print(f"GPU dostępne: Tak")
        print(f"Nazwa GPU: {torch.cuda.get_device_name(0)}")
        print(f"Pamięć GPU: {torch.cuda.get_device_properties(0).total_memory / (1024**3):.2f} GB")
    else:
        print("GPU: Niedostępne (włącz GPU w Runtime -> Change runtime type)")
except ImportError:
    print("PyTorch nie zainstalowany - nie można sprawdzić GPU")
```
