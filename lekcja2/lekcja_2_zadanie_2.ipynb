```python
import os
folder_path = "/content/drive/MyDrive/DS_Course/Lesson02"
os.makedirs(folder_path, exist_ok=True)
print("Folder został utworzony.")
```

```python
from datetime import datetime

name = "Justyna"
today = datetime.now().strftime("%Y-%m-%d")

file_path = f"{folder_path}/info.txt"

with open(file_path, "w") as file:
    file.write(f"Imię: {name}\n")
    file.write(f"Data: {today}")

print("Plik został zapisany.")
```
