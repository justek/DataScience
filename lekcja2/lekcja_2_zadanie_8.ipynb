```python
import matplotlib.pyplot as plt

# dane
devices = ["Laptop", "Desktop", "Server"]
energy = [30, 150, 400]
```

```python
# wykres
plt.figure(figsize=(6, 4))
plt.bar(devices, energy, color=["green", "orange", "red"])

# opis
plt.title("Energy Consumption of Devices")
plt.xlabel("Device")
plt.ylabel("Energy (Wh)")
```

```python
# zapis i wyswietlenie wykresu
plt.savefig("energy_comparison.png", dpi=300)
plt.show()
```
