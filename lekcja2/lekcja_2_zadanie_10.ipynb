```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
# dataset

models = ["Model A", "Model B", "Model C", "Model D", "Model E"]

data = {
    "model_name": np.repeat(models, 10),
    "training_time_min": np.random.randint(5, 120, 50),
    "cpu_percent": np.random.randint(20, 100, 50),
    "ram_gb": np.random.uniform(2, 16, 50).round(2),
    "energy_kwh": np.random.uniform(0.5, 5.0, 50).round(2)
}

df = pd.DataFrame(data)

df.head()
```

```python
# obliczanie sredniego zuzycia

average_usage = df.groupby("model_name").mean(numeric_only=True)

print(average_usage)
```

```python
# wykres

plt.figure(figsize=(8,5))

plt.bar(
    average_usage.index,
    average_usage["energy_kwh"],
    color="green"
)

plt.title("Average Energy Consumption by Model")
plt.xlabel("Model")
plt.ylabel("Energy (kWh)")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

# wnioski

- Każdy model został uruchomiony 10 razy.
- Obliczono średnie wartości czasu treningu, zużycia CPU, pamięci RAM oraz energii.
- Model z najniższą wartością energy_kwh jest najbardziej energooszczędny.
- Analiza zużycia zasobów może pomóc w wyborze modelu do wdrożenia.
