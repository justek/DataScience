```python
import numpy as np
import psutil
import time
```

```python
# wartosci przed obliczeniami
cpu_before = psutil.cpu_percent(interval=1)
ram_before = psutil.virtual_memory().used / (1024**3)
ram_before_mb = psutil.virtual_memory().used / (1024**2)
```

```python
# obliczenia
start = time.time()
matrix = np.random.rand(1000, 1000)
inverse_matrix = np.linalg.inv(matrix)
end = time.time()
```

```python
# wartosci po obliczeniach
cpu_after = psutil.cpu_percent(interval=1)
ram_after = psutil.virtual_memory().used / (1024**3)
ram_after_mb = psutil.virtual_memory().used / (1024**2)
```

```python
# raport
print("===== RAPORT ZUŻYCIA ZASOBÓW =====")
print(f"Czas wykonania: {end - start:.2f} s")
print(f"CPU przed: {cpu_before}%")
print(f"CPU po: {cpu_after}%")
print(f"RAM przed: {ram_before:.2f} GB")
print(f"RAM po: {ram_after:.2f} GB")
print(f"Różnica RAM: {ram_after - ram_before:.2f} GB")
print(f"RAM przed: {ram_before_mb:.2f} MB")
print(f"RAM po: {ram_after_mb:.2f} MB")
print(f"Różnica RAM: {ram_after_mb - ram_before_mb:.2f} MB")
```
