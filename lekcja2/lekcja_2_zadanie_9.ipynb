```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
comparison = pd.DataFrame({
    "Feature": [
        "Cost",
        "GPU Access",
        "Offline Work",
        "Time Limits",
        "Environment Control"
    ],
    "Google Colab": [
        "Free",
        "Yes",
        "No",
        "Yes",
        "Limited"
    ],
    "Jupyter Local": [
        "Free",
        "Own GPU",
        "Yes",
        "No",
        "Full"
    ]
})

comparison
```

```python
comparison.to_csv("colab_vs_jupyter.csv", index=False)
print("Plik został zapisany jako colab_vs_jupyter.csv")
```

```python
features = [
    "GPU",
    "Offline",
    "No Time Limit",
    "Full Control"
]

colab = [1, 0, 0, 0]
jupyter = [1, 1, 1, 1]

x = range(len(features))
width = 0.35

plt.figure(figsize=(8,5))

plt.bar([i - width/2 for i in x], colab, width, label="Google Colab")
plt.bar([i + width/2 for i in x], jupyter, width, label="Jupyter Local")

plt.xticks(x, features)
plt.yticks([0,1], ["No","Yes"])

plt.title("Google Colab vs Jupyter Local")
plt.ylabel("Availability")
plt.legend()

plt.show()
```
